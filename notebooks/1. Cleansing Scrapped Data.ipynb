```python
import pandas as pd

rows = []
current = None

path = f"{DATA_DIR}/1_raw/2024/google_app_reviews_1.csv"

with open(path, "r", encoding="utf-8") as f:
    for line in f:
        line = line.rstrip("\n")

        if not line.strip():
            continue  # skip empty lines safely

        parts = line.split(",")

        # detect header
        if parts[0].strip().lower() == "id":
            continue

        # detect start of a new row by numeric id
        if parts[0].strip().isdigit() and len(parts) >= 4:
            if current is not None:
                rows.append(current)

            row_id = parts[0].strip()
            app = parts[1].strip().strip('"')
            score = parts[-1].strip()

            content = ",".join(parts[2:-1]).strip().strip('"')

            current = [row_id, app, content, score]

        else:
            # continuation of multiline review text
            if current is not None:
                current[2] += " " + line.strip()
            # else: silently ignore junk before first row

    if current is not None:
        rows.append(current)

df_old_play_store_1 = pd.DataFrame(
    rows,
    columns=["id", "app", "content", "score"]
)

df_old_play_store_1
```
